```python
import altair as alt
import pandas as pd
alt.data_transformers.disable_max_rows()

df = pd.read_csv("df_full.csv")
```

# Question 1: How has clean fuel access changed over the past 20 years globally?

```python
df_fuel = df[df['clean_fuel_access_pct'].notna()].copy()

fuel_by_year = df_fuel.groupby(['location', 'year', 'Code'], as_index=False).agg({
    'clean_fuel_access_pct': 'first',
    'avg_temp_1991_2020': 'first'
})
```

```python
country_selection = alt.selection_multi(
    fields=['location'],
    bind='legend'
)

country_lines = alt.Chart(fuel_by_year).mark_line(size=2).encode(
    x=alt.X('year:O', title='Year'),
    y=alt.Y('clean_fuel_access_pct:Q', title='Clean Fuel Access (%)'),
    color=alt.Color('location:N', 
                    title='Country',
                    legend=alt.Legend(
                        orient='top',
                        columns=11,
                        symbolLimit=0,
                        labelLimit=150
                    )),
    opacity=alt.condition(country_selection, alt.value(1), alt.value(0.05)),
    strokeWidth=alt.condition(country_selection, alt.value(3), alt.value(1)),
    tooltip=['location', 'year', alt.Tooltip('clean_fuel_access_pct:Q', format='.2f', title='Clean Fuel Access (%)')]
).add_selection(
    country_selection
).transform_filter(
    country_selection
)

global_stats = alt.Chart(fuel_by_year).transform_aggregate(
    mean_access='mean(clean_fuel_access_pct)',
    sd_access='stdev(clean_fuel_access_pct)',
    groupby=['year']
).transform_calculate(
    lower='datum.mean_access - datum.sd_access',
    upper='datum.mean_access + datum.sd_access'
)

confidence_band = global_stats.mark_area(opacity=0.3, color='steelblue').encode(
    x='year:O',
    y='lower:Q',
    y2='upper:Q'
)

global_mean = global_stats.mark_line(size=5, color='darkblue').encode(
    x='year:O',
    y=alt.Y('mean_access:Q', title='Clean Fuel Access (%)'),
    tooltip=[
        alt.Tooltip('year:O', title='Year'),
        alt.Tooltip('mean_access:Q', format='.2f', title='Global Mean (%)')
    ]
)

chart1 = (confidence_band + global_mean + country_lines).properties(
    width=1300,
    height=300,
    title={
        'text': 'Global Clean Fuel Access Trends (Past 20 Years)',
        'subtitle': 'Blue band shows GLOBAL MEAN. Click legend items to filter countries'
    }
).configure_title(
    fontSize=16
).configure_legend(
    strokeColor='lightgray',
    fillColor='white',
    padding=10,
    cornerRadius=5
)

chart1
```

# Question 2: How is access to clean fuel related to mortality?

```python
df_enhanced = df.dropna(subset=['clean_fuel_access_pct', 'death_rate_per_100k']).copy()

years = sorted(df_enhanced['year'].unique())
```

```python
year_selection = alt.selection_single(
    name='SelectYear',
    fields=['year'],
    bind=alt.binding_select(options=years, name='Year: '),
    empty='all'
)

zoom = alt.selection_interval(bind='scales')

click_selection = alt.selection_multi(
    fields=['location'],
    bind='legend'
)

mortality_scatter = alt.Chart(df_enhanced).mark_circle(opacity=0.7, size=80).encode(
    x=alt.X('clean_fuel_access_pct:Q', 
            title='Clean Fuel Access (%)',
            axis=alt.Axis(minExtent=30)),
    y=alt.Y('death_rate_per_100k:Q', 
            title='Total Death Rate (per 100k population)', 
            scale=alt.Scale(type='log'),
            axis=alt.Axis(minExtent=30)),
    color=alt.condition(
        click_selection,
        alt.Color('location:N', 
                  title='Country (Click to Filter)',
                  legend=alt.Legend(
                      orient='right',
                      columns=1,
                      symbolLimit=30,
                      labelLimit=120
                  )),
        alt.value('lightgray')
    ),
    size=alt.condition(
        click_selection,
        alt.value(120),
        alt.value(40)
    ),
    tooltip=[
        alt.Tooltip('location:N', title='Location'),
        alt.Tooltip('year:O', title='Year'),
        alt.Tooltip('clean_fuel_access_pct:Q', format='.2f', title='Clean Fuel Access (%)'),
        alt.Tooltip('death_rate_per_100k:Q', format='.2f', title='Death Rate (per 100k)'),
        alt.Tooltip('avg_temp_1991_2020:Q', format='.2f', title='Avg Temperature (°C)'),
        alt.Tooltip('Health:Q', format='.2f', title='Health Score')
    ]
).add_selection(
    year_selection,
    zoom,
    click_selection
).transform_filter(
    year_selection
)

trend_line = alt.Chart(df_enhanced).mark_line(
    color='red', 
    size=4,
    strokeDash=[8, 4]
).encode(
    x=alt.X('clean_fuel_access_pct:Q'),
    y=alt.Y('death_rate_per_100k:Q')
).transform_filter(
    year_selection
).transform_filter(
    click_selection
).transform_regression(
    'clean_fuel_access_pct', 
    'death_rate_per_100k', 
    method='log'
)

mortality_chart = (mortality_scatter + trend_line).properties(
    width=1100,
    height=400,
    title={
        'text': 'Clean Fuel Access vs Mortality Rate (Interactive)',
        'subtitle': 'Select year | Click legend to filter countries | Scroll to zoom | Drag to pan'
    }
).configure_legend(
    strokeColor='lightgray',
    fillColor='white',
    padding=10,
    cornerRadius=5
).configure_title(
    fontSize=16,
    anchor='start'
)

mortality_chart
```
